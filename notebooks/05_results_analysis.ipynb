# 05 · Results analysis

RF vs CNN side by side. Plus: which gestures get confused most, and do the confusions cluster (e.g. fine finger movements vs gross wrist movements)?

```python
import sys, json
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
import pandas as pd

from src.evaluate import compare_models

rf = json.loads((ROOT / 'results' / 'metrics' / 'rf_metrics.json').read_text())
cnn = json.loads((ROOT / 'results' / 'metrics' / 'cnn_metrics.json').read_text())
comparison = compare_models(rf, cnn)
pd.DataFrame(comparison).T
```

```python
# Per-class F1, sorted by CNN performance ascending. Useful for the README writeup.
rows = []
for cls, m in cnn.items():
    if not cls.isdigit():
        continue
    rf_f1 = rf.get(cls, {}).get('f1-score', 0.0)
    rows.append({'class': int(cls), 'rf_f1': rf_f1, 'cnn_f1': m['f1-score'], 'support': m['support']})
per_class = pd.DataFrame(rows).sort_values('cnn_f1')
per_class.head(10)
```
